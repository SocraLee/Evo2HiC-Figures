```python

# This source code is licensed under the license found in the
# LICENSE file in the root directory of this source tree.
# --------------------------------------------------------
# References:
# Sagittarius to https://github.com/addiewc/Sagittarius
# --------------------------------------------------------
import sys
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D

import seaborn as sns
from plot_utils import colors as candidate_colors
from plot_utils import plot_box_with_points

import matplotlib.gridspec as gridspec

plt.style.use('seaborn-v0_8-white')
```

```python
fig = plt.figure(figsize=(8.27, 2.5))

# GridSpec
# first two rows: 4 columns; third row: 1 column
gs = gridspec.GridSpec(
    3, 4,
    figure=fig,
    height_ratios=[0.5, 2, 0.5],  # last row can be made taller
    hspace=0, wspace=0.5
)

# four panels in each of the first two rows
axes_top = []
for i in [1]:  # row
    row_axes = []
    for j in range(4):  # column
        ax = fig.add_subplot(gs[i, j])
        ax.set_xticks([]); ax.set_yticks([])
        for spine in ax.spines.values():
            spine.set_linewidth(0.5)
        row_axes.append(ax)
    axes_top.append(row_axes)

```

```python
pos0, pos1 = axes_top[0][0].get_position(), axes_top[0][3].get_position()
x = (pos0.x0+pos1.x1)/2
y = pos0.y1 + 0.15

labels = ['Evo2HiC', 'HiCNN']

# remove the old legend (avoid duplicates)
if hasattr(fig, "legend_") and fig.legend_ is not None:
    fig.legend_.remove()

handles = [Line2D([0],[0], marker='s', linestyle='', markersize=8,
                markerfacecolor=color_maps[m], markeredgewidth=0, alpha=0.8,
                label=m) for i, m in enumerate(labels)]

# separate legend at the top left
fig.legend(
    handles, labels,
    loc='center',
    bbox_to_anchor=(x, y),  # anchor slightly above the top left
    ncol=len(labels),
    frameon=False,
    fontsize=8,           # smaller font
    handletextpad=0.4,    # gap between legend marker and text
    columnspacing=3,    # gap between columns
    borderaxespad=0.1     # gap to the axes border
)
fig
```
